## scene generation function using mjSpec

```python
from spec_generate_scene import spec_generate_model
```

```python
import mujoco as mj
import mujoco_viewer

import numpy as np
import time
```

```python
multi_robot_config = {

    "scene": '../assets/floor_white_gray.xml',
    "panda": "../assets/panda/robot.xml",

    "robots":{
        "robot_1":{
        "frame": {
            "pos": [1, 0, 0],
            "quat": [0.7071, 0, 0, 0.7071] # wxyz order
        },
        "body": {
            "file": "panda",
            "body_name": "base"
        }
        },
        "robot_2":{
            "frame": {
                "pos": [0, 1, 0],
                "quat": [0, 0, 0, 1]
            },
            "body": {
                "file": "panda",
                "body_name": "base"
            }
        },
        "robot_3":{
            "frame": {
                "pos": [-1, 0, 0],
                "quat": [0.7071, 0, 0, -0.7071]
            },
            "body": {
                "file": "panda",
                "body_name": "base"
            }
        },
        "robot_4":{
            "frame": {
                "pos": [0, -1, 0],
                "quat": [1, 0, 0, 0]
            },
            "body": {
                "file": "panda",
                "body_name": "base"
            }
        }
    },

}
```

```python
model = spec_generate_model(multi_robot_config)
data = mj.MjData(model)
```

```python
""" MAIN LOOP """

# create python viewer object
viewer = mujoco_viewer.MujocoViewer(model, data)

# data reset
mj.mj_resetData(model, data)

while True:
    if viewer.is_alive:
        
        mj.mj_step(model, data)
        viewer.render()

    else:
        break

# close
viewer.close()
```
