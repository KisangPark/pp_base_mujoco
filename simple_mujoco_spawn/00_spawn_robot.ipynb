# Simple Robot Spawn
- Load robot inside mujoco
- Check with mujoco-python-viewer
    - https://github.com/rohanpsingh/mujoco-python-viewer

```python
import mujoco
import mujoco_viewer # new viewer
import numpy as np
import time

import os
```

```python
xml_path = '../asset/ur_scene.xml'
xml_abs_path = os.path.abspath(xml_path)

model = mujoco.MjModel.from_xml_path(xml_abs_path)
data = mujoco.MjData(model)
```

```python
# create mujoco-python-viewer
viewer = mujoco_viewer.MujocoViewer(model, data)

# position target
pos = [0.0] * model.nu
pos[5] = 1.6

mujoco.mj_resetData(model, data)
control_signal = pos
flag = 0
count = 0
```

```python
# simulate and render
while True:
    if viewer.is_alive:

        # set input control
        data.ctrl = control_signal 
        
        mujoco.mj_step(model, data)
        viewer.render()
    else:
        break

# close
viewer.close()
```
